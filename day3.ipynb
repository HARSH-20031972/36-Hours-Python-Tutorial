```python
count=0
user_input=input("Enter the Character=")
for char in user_input:
  if char>='1' and char<='9':
    count+=1
print(count)
```

```python
def fun(st):
  count=0
  for i in st:
    if i>='0' and i<='9':
      count=count+1
  return count
```

```python
st=input()
print(fun(st))
```

```python
user_input_str = input("Enter the rating=")
try
  user_rating = int(user_input_str)
  if user_rating >= 30 and user_rating <= 50:
    print("Average")
  elif user_rating >= 51 and user_rating <= 60:
    print("Good")
  elif user_rating >= 61 and user_rating <= 80:
    print("Excellent")
  elif user_rating >= 81 and user_rating <= 100:
    print("Outstanding")
  else:
    print("Not applicable")
except ValueError:
  print("Invalid input. Please enter a numerical rating.")
```

```python
n=input("Enter the number=")
output=""
for i in range(1,len(n),2):
  odd=int(n[i])
  odd=odd**2
  output=output+str(odd)
print(output[:4])
```

```python
def excludeddigit(num,n):
  count=0
  while(num):
    digit=num%10
    num=num//10
    if(digit!=n):
      count+=1
  return count
```

```python
num=int(input())
n=int(input())
print(excludeddigit(num,n))
```
