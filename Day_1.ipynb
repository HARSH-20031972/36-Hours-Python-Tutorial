```python
class Student:
  """Docstring of the class..."""
  def __init__(self):
    self._name="Anurag" #instance variable, protected variable
    self.roll_no=101
  def display(self):
    print("The Name is",self._name)
```

```python
s1=Student() #s is my object
print(s1._name)
print(s1.roll_no)
s1.display()
```

```python
class student:
  def __init__(self,name=".",marks=0):
    self.name=name
    self.marks=marks
  def display(self):
    print("My name is ",self.name)
    print("My Marks is ",self.marks)
```

```python
stu1=student("Anurag",500)
stu1.display()
stu2=student() #default constructor will be call
stu2.display()
```

```python
class Value:
  def __init__(self):
    self.x=10
  def modify(self):
    self.x=self.x+1
  def display(self):
    print(self.x)
```

```python
v=Value()
v2=Value()
print(v.display()) #10
v.modify() #x=11 v object calling modify
print(v2.display()) #v2 showing previous value
print(v.display())
```

```python
class Sample:
  x=10 # class variable or static variable

  @classmethod
  def modify(cls):
    cls.x+=1 #cls.x=cls.x+1
```

```python
s1=Sample()
s2=Sample()
print(s1.x)
print(s2.x)
```

```python
s1=Sample()
s2=Sample()
print(s1.x)
print(s2.x)
s1.modify()
print(s1.x)
print(s2.x)
```
