```python
class Value:
  def __init__(self):
    self.x=10
  def modify(self):
    self.x=self.x+1
  def display(self):
    print(self.x)
```

```python
v=Value()
v2=Value()
print(v.display()) #10
v.modify() #x=11 v object calling modify
print(v2.display()) #v2 showing previous value
print(v.display())
```

```python
class Sample:
  x=10 # class variable or static variable

  @classmethod
  def modify(cls):
    cls.x+=1 #cls.x=cls.x+1
```

```python
s1=Sample()
s2=Sample()
print(s1.x)
print(s2.x)
```

```python
s1=Sample()
s2=Sample()
print(s1.x)
print(s2.x)
s1.modify()
print(s1.x)
print(s2.x)
```

```python
class Student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def display(self): #method1
    print(f"My Name is {self.name} and marks i got {self.marks}")
  def calculate(self): #method2
    if(self.marks>=600):
      print("You got Grade A")
    elif(self.marks>=500):
      print("You got Grade B")
    elif(self.marks>=350):
      print("You got Grade C")
    else:
      print("you are fail")
```

```python
n=int(input("Enter the number of student"))
i=0
while(i<n):
  name=input("Enter the name of student=")
  marks=int(input("Enter the marks of student="))
  stu=Student(name,marks)
  stu.display()
  stu.calculate()
  i+=1
  print("-----------------------------------------")
```

```python
# A python program to create a bank class where deposits and and withdrawls can be handled by using instance method
```

```python
import sys
class Bank:
  def __init__(self,name):
    self.name=name
    self.balance=0.0
  def deposit(self,amt):
    self.balance+=amt
    return self.balance
  def withdraw(self,amt):
    if amt>self.balance:
      return "insufficient balance"
    self.balance-=amt
    return self.balance
```

```python
#create account
name=input("Enter the name")
b=Bank(name)
while True:
  print("d-Deposit w-withdraw e-exit")
  choice=input('your choice: ')
  if choice=='e' or choice =='E':
    exit()
  amt=float(input("Enter the amount"))
  if choice=='d' or choice=='D':
    print("Balance After Deposit",b.deposit(amt))
  elif choice=='w' or choice=='W':
    print("Balance after withdraw",b.withdraw(amt))
```

```python
#recursive call
def binary_search_recursive(arr,target,left,right):
  if left>right:
    return -1
  mid=left+(right-left)//2
  if arr[mid]==target:
    return mid
  elif arr[mid]<target:
    return binary_search_recursive(arr,target,mid+1,right)
  else:
    return binary_search_recursive(arr,target,left,mid-1)
```
